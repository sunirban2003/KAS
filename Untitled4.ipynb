```python
# Install yfinance if not already installed
# pip install yfinance pandas

import yfinance as yf
import pandas as pd

# Download daily USD/CNY data
ticker = "CNY=X"   # USD to Chinese Yuan

data = yf.download(
    ticker,
    start="2021-01-01",
    end="2026-05-01",
    interval="1d"
)

# Keep only Close price
data = data[['Close']]

# Convert index to datetime
data.index = pd.to_datetime(data.index)

# Get monthly closing price (last trading day of each month)
monthly_close = data.resample('ME').last()

# Format output
monthly_close.reset_index(inplace=True)
monthly_close.columns = ['Date', 'USD_CNY_Close']

# Keep only Jan 2021 to Apr 2026
monthly_close = monthly_close[
    (monthly_close['Date'] >= '2021-01-01') &
    (monthly_close['Date'] <= '2026-04-30')
]

# Display data
print(monthly_close)

# Save to CSV
monthly_close.to_csv("USD_CNY_Monthly_Closing_2021_2026.csv", index=False)

print("\nCSV file saved successfully!")
```

```python
# ============================================
# CAGR ANALYSIS + COMBINED EXCHANGE RATE PLOT
# ============================================

# Install packages if needed
# pip install pandas matplotlib openpyxl numpy

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ============================================
# LOAD DATA
# ============================================

file_path = "yuan_project_data.xlsx"

# Read Excel file
df = pd.read_excel(file_path)

# Display first rows
print("===== FIRST 5 ROWS =====")
print(df.head())

# Display column names
print("\n===== COLUMN NAMES =====")
print(df.columns)

# ============================================
# CONVERT MONTH COLUMN
# ============================================

# Convert Month column to datetime
df['Month'] = pd.to_datetime(df['Month'])

# Sort by Month
df = df.sort_values('Month')

# ============================================
# CAGR FUNCTION
# ============================================

def calculate_cagr(start_value, end_value, years):
    return ((end_value / start_value) ** (1 / years) - 1) * 100

# ============================================
# CALCULATE CAGR FOR ALL NUMERIC COLUMNS
# ============================================

# Select numeric columns only
numeric_cols = df.select_dtypes(include=[np.number]).columns

# Calculate total years
years = (df['Month'].iloc[-1] - df['Month'].iloc[0]).days / 365.25

print("\n===== CAGR RESULTS =====\n")

for col in numeric_cols:
    
    # Remove missing values
    temp = df[['Month', col]].dropna()
    
    start_value = temp[col].iloc[0]
    end_value = temp[col].iloc[-1]
    
    # CAGR Calculation
    cagr = calculate_cagr(start_value, end_value, years)
    
    print(f"{col}")
    print(f"Start Value : {start_value}")
    print(f"End Value   : {end_value}")
    print(f"CAGR (%)    : {round(cagr, 4)}%")
    print("-" * 40)

```

```python
# ============================================
# YEARWISE + OVERALL CAGR ANALYSIS
# FOR CURRENCY EXCHANGE RATES
# ============================================

# Install if needed
# pip install pandas numpy openpyxl

import pandas as pd
import numpy as np

# ============================================
# LOAD DATA
# ============================================

file_path = "yuan_project_data.xlsx"

df = pd.read_excel(file_path)

# Convert Month column
df['Month'] = pd.to_datetime(df['Month'])

# Sort data
df = df.sort_values('Month')

# ============================================
# SELECT CURRENCY COLUMNS
# ============================================

currency_cols = [
    'USD INR',
    'USD CNY',
    'Rs Per Yuan'
]

# ============================================
# CAGR FUNCTION
# ============================================

def calculate_cagr(start_value, end_value, years):
    return ((end_value / start_value) ** (1 / years) - 1) * 100

# ============================================
# OVERALL CAGR
# ============================================

print("\n")
print("=" * 60)
print("OVERALL CAGR ANALYSIS")
print("=" * 60)

overall_years = (
    (df['Month'].iloc[-1] - df['Month'].iloc[0]).days
    / 365.25
)

for col in currency_cols:
    
    temp = df[['Month', col]].dropna()
    
    start_value = temp[col].iloc[0]
    end_value = temp[col].iloc[-1]
    
    cagr = calculate_cagr(
        start_value,
        end_value,
        overall_years
    )
    
    print(f"\n{col}")
    print("-" * 40)
    print(f"Start Value : {start_value:.4f}")
    print(f"End Value   : {end_value:.4f}")
    print(f"Overall CAGR: {cagr:.4f}%")

# ============================================
# YEARWISE CAGR
# ============================================

print("\n\n")
print("=" * 60)
print("YEARWISE CAGR ANALYSIS")
print("=" * 60)

# Create Year column
df['Year'] = df['Month'].dt.year

for col in currency_cols:
    
    print(f"\n\n############# {col} #############")
    
    yearly_data = df[['Year', 'Month', col]].dropna()
    
    unique_years = yearly_data['Year'].unique()
    
    for year in unique_years:
        
        year_df = yearly_data[
            yearly_data['Year'] == year
        ]
        
        # Skip if insufficient data
        if len(year_df) < 2:
            continue
        
        start_value = year_df[col].iloc[0]
        end_value = year_df[col].iloc[-1]
        
        # Calculate exact year fraction
        years = (
            (year_df['Month'].iloc[-1] - year_df['Month'].iloc[0]).days
            / 365.25
        )
        
        if years == 0:
            continue
        
        cagr = calculate_cagr(
            start_value,
            end_value,
            years
        )
        
        print(f"\nYear : {year}")
        print(f"Start Value : {start_value:.4f}")
        print(f"End Value   : {end_value:.4f}")
        print(f"Yearly CAGR : {cagr:.4f}%")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.dates as mdates

# 1. Apply a modern, clean Seaborn theme
sns.set_theme(style="whitegrid")
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['DejaVu Sans', 'Arial', 'Helvetica']

# 2. Load the dataset
file_path = "yuan_project_data.xlsx"

df['Month'] = pd.to_datetime(df['Month'])

# 3. Create figure with a soft off-white background dashboard aesthetic
fig, axes = plt.subplots(3, 1, figsize=(12, 15), sharex=True, facecolor='#f8f9fa')

# Tailored color palette for distinct curves
colors = ['#1f77b4', '#00b4d8', '#e63946'] # Deep Blue, Cyan, Coral Red
columns = ['Rs Per Yuan', 'USD CNY', 'USD INR']
titles = [
    'INR per CNY Exchange Rate (Rs Per Yuan)', 
    'USD to CNY Exchange Rate', 
    'USD to INR Exchange Rate'
]

# 4. Plot each series with premium customizations
for i, col in enumerate(columns):
    ax = axes[i]
    ax.set_facecolor('#ffffff') # Crisp white plot container
    
    # Clean up outer borders (spines) for an open-grid look
    for spine in ['top', 'right', 'left', 'bottom']:
        ax.spines[spine].set_visible(False)
        
    # Plot line with optimized line weight
    ax.plot(df['Month'], df[col], color=colors[i], linewidth=2.5, zorder=3)
    
    # Add a beautiful translucent area shading underneath the curve
    ymin_boundary = df[col].min() - (df[col].max() - df[col].min()) * 0.05
    ax.fill_between(df['Month'], df[col], ymin_boundary, color=colors[i], alpha=0.08, zorder=2)
    
    # Structural titles and formatting
    ax.set_title(titles[i], fontsize=14, fontweight='bold', color='#2b2d42', pad=12, loc='left')
    ax.set_ylabel('Exchange Rate', fontsize=11, fontweight='semibold', color='#4a4e69')
    ax.grid(True, linestyle=':', alpha=0.5, color='#6c757d')
    ax.tick_params(axis='y', labelsize=10, colors='#4a4e69')

    # 5. Dynamically label Peak (Max) and Trough (Min) points
    min_idx = df[col].idxmin()
    max_idx = df[col].idxmax()
    
    # Annotate Minimum Point
    ax.scatter(df['Month'].iloc[min_idx], df[col].iloc[min_idx], color='#3a0ca3', s=65, zorder=5, edgecolors='white', linewidth=1.5)
    ax.annotate(f"Min: {df[col].iloc[min_idx]:.2f}", 
                xy=(df['Month'].iloc[min_idx], df[col].iloc[min_idx]),
                xytext=(10, -5), textcoords='offset points', fontsize=9.5, fontweight='bold', color='#3a0ca3')
                
    # Annotate Maximum Point
    ax.scatter(df['Month'].iloc[max_idx], df[col].iloc[max_idx], color='#d90429', s=65, zorder=5, edgecolors='white', linewidth=1.5)
    ax.annotate(f"Max: {df[col].iloc[max_idx]:.2f}", 
                xy=(df['Month'].iloc[max_idx], df[col].iloc[max_idx]),
                xytext=(10, 5), textcoords='offset points', fontsize=9.5, fontweight='bold', color='#d90429')

# 6. Refine the Shared X-Axis Time Format
axes[2].set_xlabel('Timeline', fontsize=13, fontweight='semibold', color='#2b2d42', labelpad=12)
axes[2].xaxis.set_major_locator(mdates.MonthLocator(interval=4)) # Shows label once every 4 months
axes[2].xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
plt.xticks(rotation=45, color='#4a4e69', fontsize=10.5)

# Overall Dashboard Title
plt.suptitle('Currency Exchange Rate Trends Analysis', fontsize=18, fontweight='bold', color='#1d3557', y=0.98)

# Polish spacing layout
plt.tight_layout(rect=[0, 0, 1, 0.96])

# Display the polished visualization
plt.show()
```
